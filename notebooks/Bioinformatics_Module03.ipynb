````python
#list the words and the number of times the word occurs in the text

with open("rosalind_ini6.txt", "r") as data:
  line = data.read()
  print(line)

  d = {}
  for word in line.split():
    if word in d:
      d[word] += 1
    else:
      d[word] = 1

print (d)

for key,value in d.items():
  print (key,value)
````
