```python
# problem 13: find a position in a genome minimizing the skew

with open('rosalind_ba1f.txt', 'r') as file:
  genome = file.read().strip()

def finding_skew(genome):
  skew_values = [0] * (len(genome) + 1)

  current_skew = 0

  for i in range(len(genome)):
    if genome[i] == "G":
      current_skew += 1
    elif genome [i] == "C":
      current_skew -= 1
    skew_values[i + 1] = current_skew

  min_skew = min(skew_values)

  min_pos = [i for i, skew in enumerate(skew_values) if skew == min_skew]

  return min_pos

print(finding_skew(genome))
```

```python
# problem 14: compute the hamming distance between two strings
with open("genome3.txt", "r") as file:
  genome3 = file.read().strip()

with open("genome4.txt", "r") as file:
  genome4 = file.read().strip()

def hamming_distance(genome3, genome4):
  distance = 0
  for i in range(min(len(genome3), len(genome4))):
    if genome3[i] != genome4[i]:
      distance += 1
  return distance

print(hamming_distance(genome3, genome4))
```
