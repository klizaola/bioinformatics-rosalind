````python
# problem 13: find a position in a genome minimizing the skew

with open('rosalind_ba1f.txt', 'r') as file:
  genome = file.read().strip()

def finding_skew(genome):
  skew_values = [0] * (len(genome) + 1)

  current_skew = 0

  for i in range(len(genome)):
    if genome[i] == "G":
      current_skew += 1
    elif genome [i] == "C":
      current_skew -= 1
    skew_values[i + 1] = current_skew

  min_skew = min(skew_values)

  min_pos = [i for i, skew in enumerate(skew_values) if skew == min_skew]

  return min_pos

print(finding_skew(genome))
````

````python
# problem 14: compute the hamming distance between two strings
with open("genome3.txt", "r") as file:
  genome3 = file.read().strip()

with open("genome4.txt", "r") as file:
  genome4 = file.read().strip()

def hamming_distance(genome3, genome4):
  distance = 0
  for i in range(min(len(genome3), len(genome4))):
    if genome3[i] != genome4[i]:
      distance += 1
  return distance

print(hamming_distance(genome3, genome4))
````

````python
pattern = "ATAAAGCCCTCC"
d = 2

alphabet = {"A", "C", "G", "T"}

def neighbors(pattern, d):
    if d == 0:
        return {pattern}

    # Initialize the set of neighbors with the original pattern (0 mismatches)
    neighborhood_set = {pattern}

    # Iterate 'd' times, each iteration increases the potential Hamming distance by 1.
    # This means after 1 iteration, we have all 1-mismatch neighbors.
    # After 2 iterations, we have all 2-mismatch neighbors (which include 1-mismatch neighbors),
    # and so on, up to 'd' mismatches.
    for _ in range(d):
        # Create a temporary set to store new neighbors generated in this iteration
        temp_neighbors = set()
        # For each pattern currently in our neighborhood (e.g., patterns with k mismatches)
        for dna in neighborhood_set:
            # Generate all 1-mismatch variations of this 'dna' pattern
            for i in range(len(dna)):
                original_char = dna[i]
                for char in alphabet:
                    if char != original_char:
                        new_pattern_list = list(dna)
                        new_pattern_list[i] = char
                        temp_neighbors.add("".join(new_pattern_list))
        # Add all newly generated neighbors to the main neighborhood set
        neighborhood_set.update(temp_neighbors)
    return neighborhood_set

# Get the neighbors
neighbor_strings = neighbors(pattern, d)

# Print each neighbor on a new line
for neighbor in sorted(list(neighbor_strings)): # Sorting for consistent output, convert to list to iterate
    print(neighbor)

# Example usage:
#pattern = "ATG"
#d = 1
#print(f"Neighbors of '{pattern}' with d={d}:\n{neighbors(pattern, d)}\n")

#pattern_2 = "ACT"
#d_2 = 2
#print(f"Neighbors of '{pattern_2}' with d={d_2}:\n{neighbors(pattern_2, d_2)}\n")
````
